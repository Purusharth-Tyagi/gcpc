```python
def admissions_agent(user_text):

    text = user_text.lower()

    # Detect B.Tech CSE
    if "b.tech" in text or "btech" in text or "बीटेक" in text:
        course = "B.Tech CSE"
    else:
        course = None

    # Detect professor request
    professor_requested = (
        "professor" in text
        or "प्रोफेसर" in text
        or "शुभांगी" in text
        or "चतुर्वेदी" in text
    )

    # Detect Darbhanga
    location_requested = (
        "darbhanga" in text
        or "दरभंगा" in text
    )

    # Build natural Hinglish response
    if course and professor_requested and location_requested:

        response = (
            "जी बिल्कुल। आपको B.Tech CSE में admission लेना है "
            "और आप Darbhanga से हैं। "
            "आप Professor Shubhangi Chaturvedi से बात करना चाहते हैं। "
            "मैं आपकी request note कर लेता हूँ।"
        )

    elif course:

        response = (
            "जी बिल्कुल। आप B.Tech CSE में admission लेना चाहते हैं। "
            "मैं आपको admission process के बारे में जानकारी देता हूँ।"
        )

    elif professor_requested:

        response = (
            "जी बिल्कुल। मैं Professor Shubhangi Chaturvedi "
            "से बात करने की आपकी request note कर लेता हूँ।"
        )

    elif location_requested:

        response = (
            "जी, मैंने note किया कि आप Darbhanga से हैं। "
            "आप admission के बारे में क्या जानना चाहते हैं?"
        )

    else:

        response = (
            "जी बिल्कुल। मैं आपकी admission में मदद कर सकता हूँ। "
            "आप अपना course या query बताइए।"
        )

    return response


# Test using the Deepgram transcript
agent_response = admissions_agent(deepgram_transcript)

print("CALLER:")
print(deepgram_transcript)

print("\nAGENT RESPONSE:")
print(agent_response)
```

```python
def final_voice(text, filename="final.wav"):
    data = {
        "text": text,
        "speaker": "celeste",
        "modelId": "arcana",
        "lang": "hin",
        "speedAlpha": 1.10
    }

    final_headers = {
        "Authorization": f"Bearer {API_KEY}",
        "Content-Type": "application/json",
        "Accept": "audio/wav"
    }

    response = requests.post(
        "https://users.rime.ai/v1/rime-tts",
        headers=final_headers,
        json=data,
        timeout=60
    )

    print("Status:", response.status_code)

    if response.status_code == 200:
        with open(filename, "wb") as f:
            f.write(response.content)

        print("Saved:", filename)
        display(Audio(filename))
    else:
        print("Rime response:", response.text)
```

```python
# RIME PHONEME OVERRIDE TEST
from google.colab import userdata
from IPython.display import Audio, display
import urllib.request
import json
import os

# Get the existing secret
RIME_API_KEY = userdata.get("RIME_API_KEY")

if not RIME_API_KEY:
    raise ValueError("RIME_API_KEY was not found in Colab Secrets.")

print("RIME API key found: True")

URL = "https://users.rime.ai/v1/rime-tts"

def generate_rime_test(text, filename, use_phoneme=False):
    payload = {
        "text": text,
        "speaker": "peak",
        "modelId": "mistv2"
    }

    if use_phoneme:
        payload["phonemizeBetweenBrackets"] = True

    headers = {
        "Accept": "audio/wav",
        "Authorization": f"Bearer {RIME_API_KEY}",
        "Content-Type": "application/json"
    }

    data = json.dumps(payload).encode("utf-8")

    request = urllib.request.Request(
        URL,
        data=data,
        headers=headers,
        method="POST"
    )

    try:
        with urllib.request.urlopen(request) as response:
            audio_data = response.read()

        with open(filename, "wb") as f:
            f.write(audio_data)

        print(f"SUCCESS: {filename}")
        print(f"Audio size: {len(audio_data)} bytes")

        display(Audio(filename))

    except Exception as e:
        print("RIME ERROR:")
        print(e)


# TEST 1 — normal pronunciation
normal_text = "Actually, custom is a word."

print("\nTEST 1 — NORMAL")
generate_rime_test(
    normal_text,
    "phoneme_test_normal.wav",
    use_phoneme=False
)


# TEST 2 — custom phoneme pronunciation
phoneme_text = "Actually, {k1Ast0xm} is a word."

print("\nTEST 2 — PHONEME OVERRIDE")
generate_rime_test(
    phoneme_text,
    "phoneme_test_override.wav",
    use_phoneme=True
)
```

```python
# FIX PHONEME TEST AUDIO FORMAT

import json
import urllib.request
from IPython.display import Audio, display
from google.colab import userdata

RIME_API_KEY = userdata.get("RIME_API_KEY")

URL = "https://users.rime.ai/v1/rime-tts"

def rime_test(text, filename):
    payload = {
        "text": text,
        "speaker": "peak",
        "modelId": "mistv2",
        "audioFormat": "wav",
        "phonemizeBetweenBrackets": True,
        "samplingRate": 22050
    }

    headers = {
        "Accept": "application/json",
        "Authorization": f"Bearer {RIME_API_KEY}",
        "Content-Type": "application/json"
    }

    request = urllib.request.Request(
        URL,
        data=json.dumps(payload).encode("utf-8"),
        headers=headers,
        method="POST"
    )

    with urllib.request.urlopen(request) as response:
        result = json.loads(response.read().decode("utf-8"))

    # Rime JSON response contains base64 audio
    import base64

    audio_data = base64.b64decode(result["audioContent"])

    with open(filename, "wb") as f:
        f.write(audio_data)

    print("SUCCESS:", filename)
    print("Size:", len(audio_data), "bytes")

    display(Audio(filename))


print("TEST 1 — NORMAL")
rime_test(
    "Actually, custom is a word.",
    "phoneme_normal.wav"
)

print("\nTEST 2 — PHONEME OVERRIDE")
rime_test(
    "Actually, {k1Ast0xm} is a word.",
    "phoneme_override.wav"
)
```
